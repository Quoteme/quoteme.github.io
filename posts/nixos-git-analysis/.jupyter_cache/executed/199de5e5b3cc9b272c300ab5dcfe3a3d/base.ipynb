```python
import os
import itertools
import functools
from datetime import datetime
import git
import plotly.graph_objects as go

# ---------------------------------------------------------------------------
# 1. Clone / update the repo using GitPython
# ---------------------------------------------------------------------------
REPO_URL = "https://github.com/quoteme/nixos.git"
REPO_DIR = "/tmp/quoteme-nixos"

repo = (
    git.Repo(REPO_DIR)
    if os.path.exists(REPO_DIR)
    else git.Repo.clone_from(REPO_URL, REPO_DIR, bare=True)
)
repo.remotes.origin.fetch("+refs/heads/*:refs/heads/*")

# ---------------------------------------------------------------------------
# 2. Parse commits (deduplicated across all refs, topo-ordered)
# ---------------------------------------------------------------------------
_seen: set[str] = set()
commits = [
    dict(
        hash  = c.hexsha,
        short = c.hexsha[:7],
        parents = [p.hexsha for p in c.parents],
        date    = datetime.fromtimestamp(c.committed_date),
        subject = c.summary,
        body    = c.message.split("\n", 1)[1].strip() if "\n" in c.message else "",
    )
    for ref in repo.refs
    for c in repo.iter_commits(ref.commit, topo_order=True)
    if c.hexsha not in _seen and not _seen.add(c.hexsha)
]

# ---------------------------------------------------------------------------
# 3. Assign branch lanes (pure fold via functools.reduce)
#    State: (lane_map, next_free_lane_id)
#    Processing order: newest → oldest, so when we see commit C we assign:
#      - C's first parent  → same lane as C  (main line stays stable)
#      - C's other parents → fresh lanes     (merged branches fan out)
# ---------------------------------------------------------------------------
def _assign_lane(
    acc: tuple[dict[str, int], int],
    commit: dict,
) -> tuple[dict[str, int], int]:
    lane_map, next_id = acc
    my_lane  = lane_map.get(commit["hash"], next_id)
    next_id2 = next_id + (0 if commit["hash"] in lane_map else 1)
    base_map = {**lane_map, commit["hash"]: my_lane}

    def _fold_parent(
        acc2: tuple[dict[str, int], int],
        ip: tuple[int, str],
    ) -> tuple[dict[str, int], int]:
        m, n = acc2
        i, p = ip
        return (m, n) if p in m else ({**m, p: (my_lane if i == 0 else n)}, n + (i != 0))

    return functools.reduce(_fold_parent, enumerate(commit["parents"]), (base_map, next_id2))

lane_map, _ = functools.reduce(_assign_lane, commits, ({}, 0))

# ---------------------------------------------------------------------------
# 4. Build edge segments (L-shaped connectors, child → parent)
# ---------------------------------------------------------------------------
hash_pos = {c["hash"]: (c["date"], lane_map.get(c["hash"], 0)) for c in commits}

edge_segments = [
    point
    for c in commits
    for p in c["parents"]
    if p in hash_pos
    for (cx, cy), (px, py) in [(hash_pos[c["hash"]], hash_pos[p])]
    for point in [(cx, cy), (px, cy), (px, py), (None, None)]
]
edge_x, edge_y = map(list, zip(*edge_segments)) if edge_segments else ([], [])

# ---------------------------------------------------------------------------
# 5. Hover labels
# ---------------------------------------------------------------------------
def _hover_label(c: dict) -> str:
    tip      = f"<b>{c['subject']}</b>" if c["subject"] else "<i>(no subject)</i>"
    body_html = "<br>" + c["body"][:300].replace("\n", "<br>") if c["body"] else ""
    return (
        f"{tip}<br>"
        f"<span style='color:#888'>{c['short']} · "
        f"{c['date'].strftime('%Y-%m-%d')}</span>"
        f"{body_html}"
    )

dates  = [c["date"]                  for c in commits]
lanes  = [lane_map.get(c["hash"], 0) for c in commits]
hover  = [_hover_label(c)            for c in commits]

# ---------------------------------------------------------------------------
# 6. Plotly figure
# ---------------------------------------------------------------------------
fig = go.Figure([
    go.Scatter(
        x=edge_x, y=edge_y,
        mode="lines",
        line=dict(width=1.5),
        hoverinfo="none",
        showlegend=False,
    ),
    go.Scatter(
        x=dates, y=lanes,
        mode="markers",
        marker=dict(
            size=9,
            color=lanes,
            colorscale="Turbo",
            line=dict(width=1),
        ),
        text=hover,
        hovertemplate="%{text}<extra></extra>",
        showlegend=False,
    ),
])

fig.update_layout(
    title=dict(text="Git history · quoteme/nixos", font=dict(size=16)),
    xaxis=dict(title="Date", showgrid=True),
    yaxis=dict(visible=False),
    height=520,
    hovermode="closest",
    plot_bgcolor="rgba(0,0,0,0)",
    paper_bgcolor="rgba(0,0,0,0)",
    margin=dict(l=20, r=20, t=50, b=40),
)

fig.show()
```
