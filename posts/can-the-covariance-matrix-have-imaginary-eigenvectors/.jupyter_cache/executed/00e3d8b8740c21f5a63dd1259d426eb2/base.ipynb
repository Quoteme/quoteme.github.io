```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

rng = np.random.default_rng(42)

# Bivariate Gaussian: std=3 along ~45° direction, std=1 orthogonal
angle = np.radians(40)
R = np.array([[np.cos(angle), -np.sin(angle)],
              [np.sin(angle),  np.cos(angle)]])
cov = R @ np.diag([9, 1]) @ R.T  # eigenvalues 9 and 1
X = rng.multivariate_normal([1, 3], cov, size=1000)

Sigma = np.cov(X.T)
eigenvalues, eigenvectors = np.linalg.eigh(Sigma)
# eigh returns ascending order; flip so largest is first
eigenvalues = eigenvalues[::-1]
eigenvectors = eigenvectors[:, ::-1]

mean = X.mean(axis=0)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# --- Left: scatter with eigenvector arrows scaled by sqrt(eigenvalue) ---
ax = axes[0]
ax.scatter(X[:, 0], X[:, 1], alpha=0.25, s=8, color="gray")
colors = ["#e63946", "#457b9d"]
for i, (lam, vec, color) in enumerate(zip(eigenvalues, eigenvectors.T, colors)):
    scale = np.sqrt(lam)
    ax.annotate(
        "", xy=mean + scale * vec, xytext=mean,
        arrowprops=dict(arrowstyle="-|>", lw=2, color=color)
    )
    ax.annotate(
        "", xy=mean - scale * vec, xytext=mean,
        arrowprops=dict(arrowstyle="-|>", lw=2, color=color)
    )
    ax.text(*(mean + (scale + 0.4) * vec),
            rf"$\lambda_{i+1}={lam:.1f}$" + f"\n$\\sigma={scale:.1f}$",
            color=color, fontsize=9, ha="center", va="center")

ax.set_aspect("equal")
ax.grid(True, linestyle=":", alpha=0.5)
ax.set_title("Principal components as eigenvectors\n(arrow length = std along PC)")
handles = [mpatches.Patch(color=c, label=f"PC {i+1} (λ={lam:.1f})")
           for i, (c, lam) in enumerate(zip(colors, eigenvalues))]
ax.legend(handles=handles, loc="upper left", fontsize=8)

# --- Right: data projected onto each PC, showing variance ---
ax2 = axes[1]
projections = (X - mean) @ eigenvectors  # shape (N, 2)
for i, (lam, color) in enumerate(zip(eigenvalues, colors)):
    proj = projections[:, i]
    ax2.hist(proj, bins=50, alpha=0.55, color=color, density=True,
             label=rf"PC {i+1}: var={proj.var():.2f}, λ={lam:.2f}")
    # overlay normal curve
    xs = np.linspace(proj.min(), proj.max(), 300)
    ax2.plot(xs, 1 / (np.sqrt(2 * np.pi * lam)) * np.exp(-xs**2 / (2 * lam)),
             color=color, lw=2)

ax2.set_xlabel("Projection value")
ax2.set_ylabel("Density")
ax2.set_title("Variance along each principal component\n= corresponding eigenvalue")
ax2.legend(fontsize=9)
ax2.grid(True, linestyle=":", alpha=0.5)

plt.tight_layout()
plt.show()
```
