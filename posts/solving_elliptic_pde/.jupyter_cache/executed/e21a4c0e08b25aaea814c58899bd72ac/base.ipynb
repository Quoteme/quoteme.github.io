```python
import numpy as np
import plotly.graph_objects as go
from functools import reduce

# ── problem parameters ────────────────────────────────────────────────────────
N = 30              # interior grid points per axis
h = 1 / (N + 1)    # grid spacing

# ── source term: two Gaussian mass concentrations ─────────────────────────────
def f(x, y):
    rho1 = 3.0 * np.exp(-60 * ((x - 0.25)**2 + (y - 0.25)**2))
    rho2 = 1.5 * np.exp(-60 * ((x - 0.70)**2 + (y - 0.65)**2))
    return 2 * np.pi * (rho1 + rho2)

# ── grid coordinates (1-indexed, interior only) ────────────────────────────────
xs = list(map(lambda i: (i + 1) * h, range(N)))
ys = list(map(lambda j: (j + 1) * h, range(N)))

# ── flat index for interior node (i, j) ──────────────────────────────────────
def idx(i, j): return i * N + j

# ── generate all (row, col, value) stencil triplets functionally ──────────────
def stencil_entries(N, h):
    diag  = [(idx(i,j), idx(i,j),    4/h**2) for i in range(N) for j in range(N)]
    left  = [(idx(i,j), idx(i,j-1), -1/h**2) for i in range(N) for j in range(1, N)]
    right = [(idx(i,j), idx(i,j+1), -1/h**2) for i in range(N) for j in range(0, N-1)]
    up    = [(idx(i,j), idx(i-1,j), -1/h**2) for i in range(1, N) for j in range(N)]
    down  = [(idx(i,j), idx(i+1,j), -1/h**2) for i in range(0, N-1) for j in range(N)]
    return diag + left + right + up + down

# ── assemble dense matrix via reduce ─────────────────────────────────────────
def add_entry(A, entry):
    r, c, v = entry
    A[r, c] += v
    return A

A = reduce(add_entry,
           stencil_entries(N, h),
           np.zeros((N * N, N * N)))

# ── right-hand side ────────────────────────────────────────────────────────────
b = np.array([f(xs[i], ys[j]) for i in range(N) for j in range(N)])

# ── solve and reshape ─────────────────────────────────────────────────────────
U = np.linalg.solve(A, b).reshape(N, N)

# ── visualise ─────────────────────────────────────────────────────────────────
X, Y = np.meshgrid(xs, ys)
fig = go.Figure(data=[
    go.Surface(x=X, y=Y, z=U.T, colorscale='Viridis',
               contours=dict(z=dict(show=True, usecolormap=True, highlightcolor="limegreen")))
])
fig.update_layout(
    title='Solution u(x,y) of −Δu = f  [from scratch, N=30]',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='u(x,y)'),
    width=750, height=600,
)
fig.show()
```

```python
import numpy as np
import scipy.sparse as sp
import scipy.sparse.linalg as spla
import plotly.graph_objects as go

# ── problem parameters ────────────────────────────────────────────────────────
N = 80
h = 1 / (N + 1)

# ── source term (same two Gaussian bumps) ─────────────────────────────────────
def f(x, y):
    rho1 = 3.0 * np.exp(-60 * ((x - 0.25)**2 + (y - 0.25)**2))
    rho2 = 1.5 * np.exp(-60 * ((x - 0.70)**2 + (y - 0.65)**2))
    return 2 * np.pi * (rho1 + rho2)

# ── 1-D tridiagonal Laplacian ─────────────────────────────────────────────────
e  = np.ones(N)
T1 = sp.spdiags([-e, 2*e, -e], [-1, 0, 1], N, N, format='csr') / h**2

# ── 2-D Laplacian via Kronecker sum ───────────────────────────────────────────
I  = sp.eye(N, format='csr')
A  = sp.kron(T1, I) + sp.kron(I, T1)          # shape (N², N²), ~5 non-zeros/row

# ── right-hand side ────────────────────────────────────────────────────────────
xs = np.array([(i + 1) * h for i in range(N)])
ys = np.array([(j + 1) * h for j in range(N)])
b  = np.array([f(xs[i], ys[j]) for i in range(N) for j in range(N)])

# ── sparse direct solve ───────────────────────────────────────────────────────
U  = spla.spsolve(A.tocsr(), b).reshape(N, N)

# ── visualise ─────────────────────────────────────────────────────────────────
X, Y = np.meshgrid(xs, ys)
fig = go.Figure(data=[
    go.Surface(x=X, y=Y, z=U.T, colorscale='Viridis',
               contours=dict(z=dict(show=True, usecolormap=True, highlightcolor="limegreen")))
])
fig.update_layout(
    title='Solution u(x,y) of −Δu = f  [scipy.sparse, N=80]',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='u(x,y)'),
    width=750, height=600,
)
fig.show()
```

```python
import numpy as np
import plotly.graph_objects as go

# ── geometry primitives ───────────────────────────────────────────────────────

def ellipsoid_mesh(cx, cy, cz, rx, ry, rz, nu=24, nv=14):
    """Triangulated ellipsoid centered at (cx,cy,cz) with radii rx,ry,rz."""
    u = np.linspace(0, 2*np.pi, nu, endpoint=False)
    v = np.linspace(0, np.pi, nv)
    U, V = np.meshgrid(u, v)
    x = (cx + rx * np.sin(V) * np.cos(U)).ravel()
    y = (cy + ry * np.sin(V) * np.sin(U)).ravel()
    z = (cz + rz * np.cos(V)).ravel()
    ti, tj, tk = [], [], []
    for i in range(nv - 1):
        for j in range(nu):
            j1 = (j + 1) % nu
            p00, p01 = i*nu+j,  i*nu+j1
            p10, p11 = (i+1)*nu+j, (i+1)*nu+j1
            ti += [p00, p10]; tj += [p10, p11]; tk += [p01, p01]
    return x, y, z, ti, tj, tk

def cylinder_mesh(cx, cy, z_bot, z_top, r, n=12):
    """Triangulated open vertical cylinder."""
    theta = np.linspace(0, 2*np.pi, n, endpoint=False)
    vx = np.tile(cx + r*np.cos(theta), 2)
    vy = np.tile(cy + r*np.sin(theta), 2)
    vz = np.concatenate([np.full(n, z_bot), np.full(n, z_top)])
    ti, tj, tk = [], [], []
    for j in range(n):
        j1 = (j + 1) % n
        ti += [j,    j1];   tj += [j1, j1+n]
        tk += [j+n,  j+n]
    return vx, vy, vz, ti, tj, tk

def merge(*meshes):
    """Merge a list of (vx,vy,vz, ti,tj,tk) meshes into one."""
    vx_all, vy_all, vz_all = [], [], []
    ti_all, tj_all, tk_all = [], [], []
    offset = 0
    for vx, vy, vz, ti, tj, tk in meshes:
        vx_all.append(vx); vy_all.append(vy); vz_all.append(vz)
        ti_all.extend(np.array(ti) + offset)
        tj_all.extend(np.array(tj) + offset)
        tk_all.extend(np.array(tk) + offset)
        offset += len(vx)
    return (np.concatenate(vx_all), np.concatenate(vy_all),
            np.concatenate(vz_all),
            np.array(ti_all), np.array(tj_all), np.array(tk_all))

# ── cow geometry ──────────────────────────────────────────────────────────────
# Coordinate system: +x = forward (nose), +z = up, +y = right flank

parts = [
    ellipsoid_mesh( 0.00,  0.00, 0.50,  1.20, 0.55, 0.48, nu=32, nv=20),  # torso
    ellipsoid_mesh(-1.00,  0.00, 0.55,  0.45, 0.50, 0.44, nu=22, nv=16),  # rump
    ellipsoid_mesh( 0.90,  0.00, 0.72,  0.32, 0.28, 0.32, nu=16, nv=12),  # neck
    ellipsoid_mesh( 1.38,  0.00, 0.68,  0.36, 0.26, 0.30, nu=22, nv=16),  # head
    ellipsoid_mesh( 1.68,  0.00, 0.58,  0.17, 0.17, 0.19, nu=14, nv=10),  # snout
    ellipsoid_mesh( 1.28, -0.28, 0.98,  0.11, 0.06, 0.10, nu=10, nv=8 ),  # ear L
    ellipsoid_mesh( 1.28,  0.28, 0.98,  0.11, 0.06, 0.10, nu=10, nv=8 ),  # ear R
    ellipsoid_mesh( 0.00,  0.00, 0.28,  0.20, 0.25, 0.08, nu=14, nv=8 ),  # udder
    cylinder_mesh(  0.60, -0.35, 0.00, 0.50, 0.09),                        # leg FL
    cylinder_mesh(  0.60,  0.35, 0.00, 0.50, 0.09),                        # leg FR
    cylinder_mesh( -0.60, -0.35, 0.00, 0.50, 0.09),                        # leg BL
    cylinder_mesh( -0.60,  0.35, 0.00, 0.50, 0.09),                        # leg BR
]

vx, vy, vz, TI, TJ, TK = merge(*parts)
verts = np.stack([vx, vy, vz], axis=1)

# ── pressure coefficient per triangle ────────────────────────────────────────
# Freestream: wind blows in –x direction (nose-on)

p0, p1, p2 = verts[TI], verts[TJ], verts[TK]
normals = np.cross(p1 - p0, p2 - p0)               # un-normalised
norms   = np.linalg.norm(normals, axis=1, keepdims=True)
norms   = np.where(norms < 1e-12, 1.0, norms)
n_hat   = normals / norms                           # unit normals

# Newtonian impact pressure
into_wind = n_hat[:, 0]                             # n · (+x) = n · (−U_inf)
Cp = np.where(into_wind > 0, 2.0 * into_wind**2, -0.15)

# ── stress concentrations at leg–body junctions ───────────────────────────────
centroids = (p0 + p1 + p2) / 3
junctions = np.array([[0.60, -0.35, 0.50],
                       [0.60,  0.35, 0.50],
                       [-0.60, -0.35, 0.50],
                       [-0.60,  0.35, 0.50]])

for jx, jy, jz in junctions:
    d2 = (centroids[:,0]-jx)**2 + (centroids[:,1]-jy)**2 + (centroids[:,2]-jz)**2
    Cp = np.maximum(Cp, 1.8 * np.exp(-18 * d2))    # peak stress at junction

# ── plotly mesh3d ─────────────────────────────────────────────────────────────
colorscale = [
    [0.00, '#2166ac'],   # suction  (blue)
    [0.10, '#74add1'],
    [0.25, '#e0f3f8'],
    [0.45, '#ffffbf'],   # neutral  (yellow)
    [0.65, '#fee090'],
    [0.80, '#f46d43'],
    [1.00, '#a50026'],   # stagnation / stress (red)
]

fig = go.Figure(data=[go.Mesh3d(
    x=vx, y=vy, z=vz,
    i=TI, j=TJ, k=TK,
    intensity=Cp,
    intensitymode='cell',
    colorscale=colorscale,
    cmin=-0.15, cmax=1.8,
    colorbar=dict(title='Cp / stress', thickness=16,
                  tickvals=[-0.15, 0, 0.5, 1.0, 1.8],
                  ticktext=['-0.15 (suction)', '0', '0.5', '1.0 (stagnation)', '1.8 (junction)']),
    lighting=dict(ambient=0.45, diffuse=0.85, specular=0.25, roughness=0.4),
    lightposition=dict(x=300, y=200, z=400),
    showscale=True,
)])

fig.update_layout(
    title=dict(text=(
        'Aerodynamics of a cow — C<sub>p</sub> surface map<br>'
        '<sup>Newtonian impact model · wind from +x · FEM-style stress concentrations at leg–body junctions</sup>'
    ), x=0.5),
    scene=dict(
        xaxis_title='x → forward (nose)',
        yaxis_title='y → lateral',
        zaxis_title='z → up',
        aspectmode='data',
        camera=dict(eye=dict(x=1.6, y=-1.6, z=0.9)),
        annotations=[
            dict(x=1.68, y=0, z=0.58, text='Stagnation<br>(Cp ≈ 2)',
                 showarrow=True, arrowhead=2, font=dict(size=11, color='#a50026')),
            dict(x=0.0,  y=0.55, z=0.9, text='Suction zone<br>(Cp < 0)',
                 showarrow=True, arrowhead=2, font=dict(size=11, color='#2166ac')),
            dict(x=0.60, y=-0.35, z=0.50, text='Stress<br>concentration',
                 showarrow=True, arrowhead=2, font=dict(size=11, color='#a50026')),
        ],
    ),
    width=860, height=680,
)

fig.show()
```
