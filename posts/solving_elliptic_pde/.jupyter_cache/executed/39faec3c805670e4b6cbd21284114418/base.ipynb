```python
import numpy as np
import plotly.graph_objects as go
from functools import reduce

# ── problem parameters ────────────────────────────────────────────────────────
N = 30              # interior grid points per axis
h = 1 / (N + 1)    # grid spacing

# ── source term: two Gaussian mass concentrations ─────────────────────────────
def f(x, y):
    rho1 = 3.0 * np.exp(-60 * ((x - 0.25)**2 + (y - 0.25)**2))
    rho2 = 1.5 * np.exp(-60 * ((x - 0.70)**2 + (y - 0.65)**2))
    return 2 * np.pi * (rho1 + rho2)

# ── grid coordinates (1-indexed, interior only) ────────────────────────────────
xs = list(map(lambda i: (i + 1) * h, range(N)))
ys = list(map(lambda j: (j + 1) * h, range(N)))

# ── flat index for interior node (i, j) ──────────────────────────────────────
def idx(i, j): return i * N + j

# ── generate all (row, col, value) stencil triplets functionally ──────────────
def stencil_entries(N, h):
    diag  = [(idx(i,j), idx(i,j),    4/h**2) for i in range(N) for j in range(N)]
    left  = [(idx(i,j), idx(i,j-1), -1/h**2) for i in range(N) for j in range(1, N)]
    right = [(idx(i,j), idx(i,j+1), -1/h**2) for i in range(N) for j in range(0, N-1)]
    up    = [(idx(i,j), idx(i-1,j), -1/h**2) for i in range(1, N) for j in range(N)]
    down  = [(idx(i,j), idx(i+1,j), -1/h**2) for i in range(0, N-1) for j in range(N)]
    return diag + left + right + up + down

# ── assemble dense matrix via reduce ─────────────────────────────────────────
def add_entry(A, entry):
    r, c, v = entry
    A[r, c] += v
    return A

A = reduce(add_entry,
           stencil_entries(N, h),
           np.zeros((N * N, N * N)))

# ── right-hand side ────────────────────────────────────────────────────────────
b = np.array([f(xs[i], ys[j]) for i in range(N) for j in range(N)])

# ── solve and reshape ─────────────────────────────────────────────────────────
U = np.linalg.solve(A, b).reshape(N, N)

# ── visualise ─────────────────────────────────────────────────────────────────
X, Y = np.meshgrid(xs, ys)
fig = go.Figure(data=[
    go.Surface(x=X, y=Y, z=U.T, colorscale='Viridis',
               contours=dict(z=dict(show=True, usecolormap=True, highlightcolor="limegreen")))
])
fig.update_layout(
    title='Solution u(x,y) of −Δu = f  [from scratch, N=30]',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='u(x,y)'),
    width=750, height=600,
)
fig.show()
```

```python
import numpy as np
import scipy.sparse as sp
import scipy.sparse.linalg as spla
import plotly.graph_objects as go

# ── problem parameters ────────────────────────────────────────────────────────
N = 80
h = 1 / (N + 1)

# ── source term (same two Gaussian bumps) ─────────────────────────────────────
def f(x, y):
    rho1 = 3.0 * np.exp(-60 * ((x - 0.25)**2 + (y - 0.25)**2))
    rho2 = 1.5 * np.exp(-60 * ((x - 0.70)**2 + (y - 0.65)**2))
    return 2 * np.pi * (rho1 + rho2)

# ── 1-D tridiagonal Laplacian ─────────────────────────────────────────────────
e  = np.ones(N)
T1 = sp.spdiags([-e, 2*e, -e], [-1, 0, 1], N, N, format='csr') / h**2

# ── 2-D Laplacian via Kronecker sum ───────────────────────────────────────────
I  = sp.eye(N, format='csr')
A  = sp.kron(T1, I) + sp.kron(I, T1)          # shape (N², N²), ~5 non-zeros/row

# ── right-hand side ────────────────────────────────────────────────────────────
xs = np.array([(i + 1) * h for i in range(N)])
ys = np.array([(j + 1) * h for j in range(N)])
b  = np.array([f(xs[i], ys[j]) for i in range(N) for j in range(N)])

# ── sparse direct solve ───────────────────────────────────────────────────────
U  = spla.spsolve(A.tocsr(), b).reshape(N, N)

# ── visualise ─────────────────────────────────────────────────────────────────
X, Y = np.meshgrid(xs, ys)
fig = go.Figure(data=[
    go.Surface(x=X, y=Y, z=U.T, colorscale='Viridis',
               contours=dict(z=dict(show=True, usecolormap=True, highlightcolor="limegreen")))
])
fig.update_layout(
    title='Solution u(x,y) of −Δu = f  [scipy.sparse, N=80]',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='u(x,y)'),
    width=750, height=600,
)
fig.show()
```

```python
import trimesh
import numpy as np
import plotly.graph_objects as go

# ── Load + centre the cow mesh ─────────────────────────────────────────────────
mesh = trimesh.load('../../assets/lowpolycow/cow.obj', force='mesh')
mesh.apply_transform(trimesh.transformations.rotation_matrix(-np.pi / 2, [0, 1, 0]))
mesh.vertices -= mesh.centroid

verts = mesh.vertices   # (V, 3)
faces = mesh.faces      # (F, 3)

# ── Aerodynamic setup ──────────────────────────────────────────────────────────
# Wind blows in the +z direction (head-on from the front of the cow).
wind_dir = np.array([0., 0., 1.])
normals  = mesh.face_normals               # (F, 3), unit vectors

# Modified Newtonian pressure coefficient (inviscid, attached-flow approximation)
#   Cp =  2 cos²θ       for windward faces (normal partly facing flow)
#   Cp = -0.4 |cosθ|^0.8 for leeward faces (suction)
cos_theta = normals @ wind_dir             # (F,)  in [-1, 1]
windward  = np.maximum(0.0,  cos_theta)
leeward   = np.maximum(0.0, -cos_theta)
Cp_face   = 2.0 * windward**2 - 0.4 * leeward**0.8

# ── Average face Cp to vertices (vectorised) ───────────────────────────────────
Cp_vert  = np.zeros(len(verts))
vert_cnt = np.zeros(len(verts))
np.add.at(Cp_vert,  faces.ravel(), np.repeat(Cp_face, 3))
np.add.at(vert_cnt, faces.ravel(), 1)
Cp_vert /= np.maximum(vert_cnt, 1)

# ── Plot ───────────────────────────────────────────────────────────────────────
fig = go.Figure(data=[go.Mesh3d(
    x=verts[:,0], y=verts[:,1], z=verts[:,2],
    i=faces[:,0], j=faces[:,1], k=faces[:,2],
    intensity=Cp_vert,
    colorscale='RdBu_r',
    cmin=Cp_vert.min(), cmax=Cp_vert.max(),
    colorbar=dict(title='Cp', thickness=18, len=0.7),
    lighting=dict(ambient=0.5, diffuse=0.8, specular=0.2, roughness=0.5),
    flatshading=False,
    name='Pressure coefficient',
    showscale=True,
)])
fig.update_layout(
    title='Aerodynamic surface pressure Cp on a cow<br>'
          '<sub>Wind → +z (head-on). Red = high pressure (stagnation), '
          'Blue = suction (leeward)</sub>',
    scene=dict(
        xaxis_title='x', yaxis_title='y', zaxis_title='z',
        aspectmode='data',
        camera=dict(eye=dict(x=1.6, y=0.8, z=-1.0)),
    ),
    width=800, height=650,
)
fig.show()
```

```python
import trimesh
import numpy as np
import plotly.graph_objects as go

# ── Re-load mesh (cell is self-contained) ─────────────────────────────────────
mesh = trimesh.load('../../assets/lowpolycow/cow.obj', force='mesh')
mesh.apply_transform(trimesh.transformations.rotation_matrix(-np.pi / 2, [0, 0, 1]))
mesh.vertices -= mesh.centroid
verts, faces = mesh.vertices, mesh.faces

# ── Aerodynamic Cp (same as above) ────────────────────────────────────────────
wind_dir  = np.array([0., 0., 1.])
cos_theta = mesh.face_normals @ wind_dir
windward  = np.maximum(0.0,  cos_theta)
leeward   = np.maximum(0.0, -cos_theta)
Cp_face   = 2.0 * windward**2 - 0.4 * leeward**0.8
Cp_vert   = np.zeros(len(verts)); vert_cnt = np.zeros(len(verts))
np.add.at(Cp_vert,  faces.ravel(), np.repeat(Cp_face, 3))
np.add.at(vert_cnt, faces.ravel(), 1)
Cp_vert /= np.maximum(vert_cnt, 1)
Cp_pos   = np.maximum(0.0, Cp_vert)                        # compressive only

# ── Gaussian curvature via angular defect ─────────────────────────────────────
defects  = trimesh.curvature.vertex_defects(mesh)          # radians; large at sharp corners
kappa    = np.abs(defects)
kappa_n  = (kappa - kappa.min()) / (kappa.max() - kappa.min() + 1e-12)

# ── Normalise Cp and build stress index ───────────────────────────────────────
Cp_n     = (Cp_pos - Cp_pos.min()) / (Cp_pos.max() - Cp_pos.min() + 1e-12)
alpha    = 0.6
sigma    = alpha * Cp_n + (1 - alpha) * kappa_n
sigma_n  = (sigma - sigma.min()) / (sigma.max() - sigma.min())

# ── Identify problem zones (top 8 % stress) ────────────────────────────────────
threshold   = np.percentile(sigma_n, 92)
hot_mask    = sigma_n > threshold
hot_verts   = verts[hot_mask]

# Cluster the hot vertices into labelled zones using simple x-z binning
def zone_centroid(mask, label):
    pts = verts[mask]
    return pts.mean(axis=0), label

# Manually curate the most representative hotspot regions
zones = {
    'Neck / head junction\n(notch effect)'   : (verts[:,2] > 0.35) & hot_mask,
    'Forelegs\n(stress concentration)'       : (verts[:,1] < -0.5) & (verts[:,2] > 0.0) & hot_mask,
    'Hindlegs\n(stress concentration)'       : (verts[:,1] < -0.5) & (verts[:,2] < 0.0) & hot_mask,
    'Dorsal ridge\n(curvature peak)'         : (verts[:,1] >  0.6) & hot_mask,
    'Udder / ventral\n(max aerodrag zone)'   : (verts[:,1] < -0.3) & (np.abs(verts[:,0]) < 0.2) & hot_mask,
}

# ── Main mesh plot ─────────────────────────────────────────────────────────────
traces = [go.Mesh3d(
    x=verts[:,0], y=verts[:,1], z=verts[:,2],
    i=faces[:,0], j=faces[:,1], k=faces[:,2],
    intensity=sigma_n,
    colorscale=[
        [0.00, '#2166ac'], [0.40, '#74add1'],
        [0.65, '#fee090'], [0.82, '#f46d43'],
        [1.00, '#a50026'],
    ],
    cmin=0, cmax=1,
    colorbar=dict(
        title='Stress index σ*',
        tickvals=[0, 0.5, 1],
        ticktext=['Low', 'Med', 'High'],
        thickness=18, len=0.65,
    ),
    lighting=dict(ambient=0.45, diffuse=0.85, specular=0.3, roughness=0.4),
    flatshading=False,
    name='FEM stress index',
    showscale=True,
    opacity=1.0,
)]

# ── Add problem-zone markers and annotation cones ─────────────────────────────
colours = ['#ff0000', '#ff7700', '#cc00cc', '#007700', '#0000cc']
for (label, mask), col in zip(zones.items(), colours):
    if mask.sum() == 0:
        continue
    centroid = verts[mask].mean(axis=0)
    # sphere of small markers at hotspot vertices
    traces.append(go.Scatter3d(
        x=verts[mask, 0], y=verts[mask, 1], z=verts[mask, 2],
        mode='markers',
        marker=dict(size=2.5, color=col, opacity=0.6),
        name=label.replace('\n', ' '),
        showlegend=True,
    ))
    # annotation cone pointing towards centroid
    offset = centroid + np.array([0.0, 0.35, 0.0])
    traces.append(go.Scatter3d(
        x=[offset[0], centroid[0]],
        y=[offset[1], centroid[1]],
        z=[offset[2], centroid[2]],
        mode='lines+text',
        line=dict(color=col, width=3),
        text=['', label],
        textposition='top center',
        textfont=dict(size=9, color=col),
        showlegend=False,
    ))

fig = go.Figure(data=traces)
fig.update_layout(
    title='FEM stress index σ* on a cow — problem zones highlighted<br>'
          '<sub>Wind → +z. Red = critical stress. Coloured markers = identified risk zones.</sub>',
    scene=dict(
        xaxis_title='x', yaxis_title='y (up)', zaxis_title='z (flow)',
        aspectmode='data',
        camera=dict(eye=dict(x=1.4, y=0.9, z=-0.8)),
    ),
    legend=dict(x=0.01, y=0.99, bgcolor='rgba(255,255,255,0.7)', borderwidth=1),
    width=850, height=700,
)
fig.show()
```
