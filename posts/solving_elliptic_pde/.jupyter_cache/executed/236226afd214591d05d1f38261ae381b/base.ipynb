```python
import numpy as np
import plotly.graph_objects as go
from functools import reduce

# ── problem parameters ────────────────────────────────────────────────────────
N = 30              # interior grid points per axis
h = 1 / (N + 1)    # grid spacing

# ── source term: two Gaussian mass concentrations ─────────────────────────────
def f(x, y):
    rho1 = 3.0 * np.exp(-60 * ((x - 0.25)**2 + (y - 0.25)**2))
    rho2 = 1.5 * np.exp(-60 * ((x - 0.70)**2 + (y - 0.65)**2))
    return 2 * np.pi * (rho1 + rho2)

# ── grid coordinates (1-indexed, interior only) ────────────────────────────────
xs = list(map(lambda i: (i + 1) * h, range(N)))
ys = list(map(lambda j: (j + 1) * h, range(N)))

# ── flat index for interior node (i, j) ──────────────────────────────────────
def idx(i, j): return i * N + j

# ── generate all (row, col, value) stencil triplets functionally ──────────────
def stencil_entries(N, h):
    diag  = [(idx(i,j), idx(i,j),    4/h**2) for i in range(N) for j in range(N)]
    left  = [(idx(i,j), idx(i,j-1), -1/h**2) for i in range(N) for j in range(1, N)]
    right = [(idx(i,j), idx(i,j+1), -1/h**2) for i in range(N) for j in range(0, N-1)]
    up    = [(idx(i,j), idx(i-1,j), -1/h**2) for i in range(1, N) for j in range(N)]
    down  = [(idx(i,j), idx(i+1,j), -1/h**2) for i in range(0, N-1) for j in range(N)]
    return diag + left + right + up + down

# ── assemble dense matrix via reduce ─────────────────────────────────────────
def add_entry(A, entry):
    r, c, v = entry
    A[r, c] += v
    return A

A = reduce(add_entry,
           stencil_entries(N, h),
           np.zeros((N * N, N * N)))

# ── right-hand side ────────────────────────────────────────────────────────────
b = np.array([f(xs[i], ys[j]) for i in range(N) for j in range(N)])

# ── solve and reshape ─────────────────────────────────────────────────────────
U = np.linalg.solve(A, b).reshape(N, N)

# ── visualise ─────────────────────────────────────────────────────────────────
X, Y = np.meshgrid(xs, ys)
fig = go.Figure(data=[
    go.Surface(x=X, y=Y, z=U.T, colorscale='Viridis',
               contours=dict(z=dict(show=True, usecolormap=True, highlightcolor="limegreen")))
])
fig.update_layout(
    title='Solution u(x,y) of −Δu = f  [from scratch, N=30]',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='u(x,y)'),
    width=750, height=600,
)
fig.show()
```

```python
import numpy as np
import scipy.sparse as sp
import scipy.sparse.linalg as spla
import plotly.graph_objects as go

# ── problem parameters ────────────────────────────────────────────────────────
N = 80
h = 1 / (N + 1)

# ── source term (same two Gaussian bumps) ─────────────────────────────────────
def f(x, y):
    rho1 = 3.0 * np.exp(-60 * ((x - 0.25)**2 + (y - 0.25)**2))
    rho2 = 1.5 * np.exp(-60 * ((x - 0.70)**2 + (y - 0.65)**2))
    return 2 * np.pi * (rho1 + rho2)

# ── 1-D tridiagonal Laplacian ─────────────────────────────────────────────────
e  = np.ones(N)
T1 = sp.spdiags([-e, 2*e, -e], [-1, 0, 1], N, N, format='csr') / h**2

# ── 2-D Laplacian via Kronecker sum ───────────────────────────────────────────
I  = sp.eye(N, format='csr')
A  = sp.kron(T1, I) + sp.kron(I, T1)          # shape (N², N²), ~5 non-zeros/row

# ── right-hand side ────────────────────────────────────────────────────────────
xs = np.array([(i + 1) * h for i in range(N)])
ys = np.array([(j + 1) * h for j in range(N)])
b  = np.array([f(xs[i], ys[j]) for i in range(N) for j in range(N)])

# ── sparse direct solve ───────────────────────────────────────────────────────
U  = spla.spsolve(A.tocsr(), b).reshape(N, N)

# ── visualise ─────────────────────────────────────────────────────────────────
X, Y = np.meshgrid(xs, ys)
fig = go.Figure(data=[
    go.Surface(x=X, y=Y, z=U.T, colorscale='Viridis',
               contours=dict(z=dict(show=True, usecolormap=True, highlightcolor="limegreen")))
])
fig.update_layout(
    title='Solution u(x,y) of −Δu = f  [scipy.sparse, N=80]',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='u(x,y)'),
    width=750, height=600,
)
fig.show()
```

```python
import numpy as np
from scipy.spatial import Delaunay
import scipy.sparse as sp
import scipy.sparse.linalg as spla
import plotly.graph_objects as go

# ── Material  (structural steel S235 / ASTM A36) ──────────────────────────────
E       = 200_000.0   # Young's modulus [MPa]
nu      = 0.30        # Poisson's ratio
sigma0  = 100.0       # Applied far-field tension [MPa]
sigma_y = 250.0       # Yield stress [MPa]

# ── Plate geometry ─────────────────────────────────────────────────────────────
L = 1.0    # plate half-width = half-height [m]
R = 0.25   # hole radius [m]

# ── Build Delaunay mesh with a circular hole ───────────────────────────────────
ng  = 28                                     # coarse grid resolution
xg  = np.linspace(-L, L, ng)
yg  = np.linspace(-L, L, ng)
Xg, Yg = np.meshgrid(xg, yg)
pts = np.column_stack([Xg.ravel(), Yg.ravel()])
pts = pts[pts[:, 0]**2 + pts[:, 1]**2 > (R * 1.05)**2]   # remove hole interior

# Dense ring on the hole boundary (needed for accurate stress at the hot-spot)
th     = np.linspace(0, 2*np.pi, 64, endpoint=False)
pts_ring = np.column_stack([R * np.cos(th), R * np.sin(th)])
pts    = np.vstack([pts, pts_ring])

tri    = Delaunay(pts)
elems  = tri.simplices
# Drop triangles whose centroid falls inside the hole
cx, cy = pts[elems, 0].mean(1), pts[elems, 1].mean(1)
elems  = elems[(cx**2 + cy**2) > R**2]

nodes   = pts
n_nodes = len(nodes)
n_dof   = 2 * n_nodes

# ── Plane-stress constitutive matrix ──────────────────────────────────────────
D = E / (1 - nu**2) * np.array([
    [1,  nu, 0          ],
    [nu, 1,  0          ],
    [0,  0,  (1-nu)/2   ],
])

# ── Assemble global stiffness K ────────────────────────────────────────────────
rows_k, cols_k, vals_k = [], [], []

for tri_nd in elems:
    xc, yc = nodes[tri_nd, 0], nodes[tri_nd, 1]
    xi, xj, xm = xc;  yi, yj, ym = yc

    A2 = (xj - xi)*(ym - yi) - (xm - xi)*(yj - yi)   # twice the element area
    if abs(A2) < 1e-15: continue
    if A2 < 0:                                          # enforce CCW orientation
        tri_nd = tri_nd[[0, 2, 1]];  xc, yc = nodes[tri_nd, 0], nodes[tri_nd, 1]
        xi, xj, xm = xc;  yi, yj, ym = yc;  A2 = -A2

    bi, bj, bm = yj - ym, ym - yi, yi - yj             # ∂N/∂x coefficients × A2
    ci, cj, cm = xm - xj, xi - xm, xj - xi             # ∂N/∂y coefficients × A2

    B  = np.array([[bi, 0,  bj, 0,  bm, 0 ],
                   [0,  ci, 0,  cj, 0,  cm],
                   [ci, bi, cj, bj, cm, bm]]) / A2      # strain-displacement matrix

    Ke = (A2 / 2) * B.T @ D @ B                         # 6×6 element stiffness

    dofs = np.empty(6, int)
    dofs[0::2] = 2 * tri_nd;  dofs[1::2] = 2 * tri_nd + 1
    ii, jj = np.meshgrid(dofs, dofs, indexing='ij')
    rows_k.extend(ii.ravel());  cols_k.extend(jj.ravel());  vals_k.extend(Ke.ravel())

K = sp.csr_matrix((vals_k, (rows_k, cols_k)), shape=(n_dof, n_dof)).tolil()

# ── Load: uniform traction σ₀ on the right edge ───────────────────────────────
tol   = 2 * L / (ng - 1) * 0.6          # ≈ one grid spacing
F     = np.zeros(n_dof)
right = np.where(nodes[:, 0] > L - tol)[0]
F[2 * right] = sigma0 * 2 * L / len(right)   # total force shared equally

# ── Dirichlet BCs: fix both DOFs on left edge (penalty method) ────────────────
left = np.where(nodes[:, 0] < -L + tol)[0]
pen  = 1e14
for nd in left:
    K[2*nd,   2*nd  ] += pen
    K[2*nd+1, 2*nd+1] += pen

K = K.tocsr()

# ── Linear solve ──────────────────────────────────────────────────────────────
u  = spla.spsolve(K, F)
ux = u[0::2];  uy = u[1::2]

# ── Element von Mises stress ──────────────────────────────────────────────────
vm_elem = np.zeros(len(elems))

for k, tri_nd in enumerate(elems):
    xc, yc = nodes[tri_nd, 0], nodes[tri_nd, 1]
    xi, xj, xm = xc;  yi, yj, ym = yc
    A2 = (xj-xi)*(ym-yi) - (xm-xi)*(yj-yi)
    if abs(A2) < 1e-15: continue
    if A2 < 0: A2 = -A2

    bi, bj, bm = yj-ym, ym-yi, yi-yj
    ci, cj, cm = xm-xj, xi-xm, xj-xi
    B  = np.array([[bi, 0,  bj, 0,  bm, 0 ],
                   [0,  ci, 0,  cj, 0,  cm],
                   [ci, bi, cj, bj, cm, bm]]) / A2
    ue = np.array([ux[tri_nd[0]], uy[tri_nd[0]],
                   ux[tri_nd[1]], uy[tri_nd[1]],
                   ux[tri_nd[2]], uy[tri_nd[2]]])
    s  = D @ B @ ue                                        # [σx, σy, τxy]
    vm_elem[k] = np.sqrt(s[0]**2 + s[1]**2 - s[0]*s[1] + 3*s[2]**2)

# ── Node-averaged von Mises (Zienkiewicz–Zhu smoothing) ───────────────────────
vm_node = np.zeros(n_nodes);  cnt = np.zeros(n_nodes)
for k, tri_nd in enumerate(elems):
    vm_node[tri_nd] += vm_elem[k];  cnt[tri_nd] += 1
vm_node = np.where(cnt > 0, vm_node / cnt, 0.0)

# ── Deformed shape (scaled for visibility) ────────────────────────────────────
scale  = 0.25 * L / (np.abs(ux).max() + 1e-14)
x_def  = nodes[:, 0] + scale * ux
y_def  = nodes[:, 1] + scale * uy
z_def  = vm_node / sigma0         # normalised: Kirsch peak ≈ 3

I_t, J_t, K_t = elems[:, 0], elems[:, 1], elems[:, 2]
yielded  = vm_elem > sigma_y
n_yield  = int(yielded.sum())

# ── Plotly 3-D plot ───────────────────────────────────────────────────────────
fig = go.Figure()

# Stress surface (z = σ_vm / σ₀, colour = same)
fig.add_trace(go.Mesh3d(
    x=x_def, y=y_def, z=z_def,
    i=I_t, j=J_t, k=K_t,
    intensity=vm_node / sigma0,
    colorscale=[
        [0.00, '#1a9850'],   # green  – well within elastic range
        [0.40, '#fee08b'],   # yellow – moderate
        [0.75, '#f46d43'],   # orange – approaching yield
        [1.00, '#d73027'],   # red    – at / above yield
    ],
    cmin=0.0, cmax=3.5,
    colorbar=dict(title='σ_vm / σ₀', thickness=16, len=0.75),
    name='von Mises stress',
    showscale=True,
    flatshading=False,
    lighting=dict(ambient=0.55, diffuse=0.85, roughness=0.4),
    lightposition=dict(x=100, y=200, z=800),
))

# Highlight yielded elements (σ_vm > σ_yield)
if n_yield > 0:
    iy = elems[yielded, 0];  jy = elems[yielded, 1];  ky = elems[yielded, 2]
    fig.add_trace(go.Mesh3d(
        x=x_def, y=y_def, z=z_def + 0.03,   # slight offset so it renders on top
        i=iy, j=jy, k=ky,
        color='rgba(200, 0, 40, 0.85)',
        name=f'⚠ Yielded  ({n_yield} elements, σ_vm > {sigma_y:.0f} MPa)',
        showscale=False,
        flatshading=True,
    ))

fig.update_layout(
    title=(
        f'FEM — Steel plate with hole under uniaxial tension σ₀ = {sigma0:.0f} MPa<br>'
        f'<sup>Deformed shape (×{scale:.0f}), colored by normalised von Mises stress'
        f' — {n_yield} elements above yield (σ_y = {sigma_y:.0f} MPa)</sup>'
    ),
    scene=dict(
        xaxis_title='x [m]',
        yaxis_title='y [m]',
        zaxis_title='σ_vm / σ₀',
        aspectratio=dict(x=1.0, y=1.0, z=0.55),
        camera=dict(eye=dict(x=1.6, y=-1.6, z=1.2)),
    ),
    legend=dict(x=0.01, y=0.98, bgcolor='rgba(255,255,255,0.7)',
                bordercolor='#ccc', borderwidth=1),
    width=850, height=680,
)
fig.show()

print(f"Peak σ_vm  = {vm_node.max():.1f} MPa")
print(f"Kirsch ref = {3*sigma0:.1f} MPa  (= 3 · σ₀ for an infinite plate)")
print(f"SCF (numerical / analytic) = {vm_node.max() / (3*sigma0):.3f}")
```
