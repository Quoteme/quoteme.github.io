```python
import numpy as np
import plotly.graph_objects as go

# Parameters
N = 10
h = 1 / (N + 1)

# Unit square in (x,y) plane
square = ([0, 1, 1, 0, 0], [0, 0, 1, 1, 0], [0]*5)

# Interior grid points (blue)
grid_x = [i*h for i in range(1, N+1) for j in range(1, N+1)]
grid_y = [j*h for i in range(1, N+1) for j in range(1, N+1)]
grid_z = [0] * len(grid_x)

# Boundary points (red)
boundary = [(x,y,z) 
  for (x,i) in [ (i*h,i) for i in range(N+2) ] 
  for (y,j) in [ (i*h,i) for i in range(N+2) ] 
  for z in [0]
  if i == 0 or j == 0 or i == N+1 or j == N+1
]

# Create figure
fig = go.Figure()

# Add unit square, interior grid points, and boundary points
fig.add_trace(go.Scatter3d(x=square[0], y=square[1], z=square[2], mode='lines',
                           line=dict(color='black', width=4), name='Unit Square'))
fig.add_trace(go.Scatter3d(x=grid_x, y=grid_y, z=grid_z, mode='markers',
                           marker=dict(size=5, color='blue'), name=f'Interior (N={N})'))
fig.add_trace(go.Scatter3d(
  x=[v[0] for v in boundary],
  y=[v[1] for v in boundary],
  z=[v[2] for v in boundary],
  mode='markers',
  marker=dict(size=5, color='red'),
  name='Boundary'))

# Update layout
fig.update_layout(
    title=f'Unit Square in (x,y) Plane with Grid Points (N={N}, h={h:.4f})',
    scene=dict(xaxis_title='X', yaxis_title='Y', zaxis_title='Z', aspectmode='cube'),
    showlegend=True, width=800, height=800
)

fig.show()
```

```python
import numpy as np
import plotly.graph_objects as go
from functools import reduce

# ── problem parameters ────────────────────────────────────────────────────────
N = 30              # interior grid points per axis
h = 1 / (N + 1)    # grid spacing

# ── source term: two Gaussian mass concentrations ─────────────────────────────
def f(x, y):
    rho1 = 3.0 * np.exp(-60 * ((x - 0.25)**2 + (y - 0.25)**2))
    rho2 = 1.5 * np.exp(-60 * ((x - 0.70)**2 + (y - 0.65)**2))
    return 2 * np.pi * (rho1 + rho2)

# ── grid coordinates (1-indexed, interior only) ────────────────────────────────
xs = list(map(lambda i: (i + 1) * h, range(N)))
ys = list(map(lambda j: (j + 1) * h, range(N)))

# ── flat index for interior node (i, j) ──────────────────────────────────────
def idx(i, j): return i * N + j

# ── generate all (row, col, value) stencil triplets functionally ──────────────
def stencil_entries(N, h):
    diag  = [(idx(i,j), idx(i,j),    4/h**2) for i in range(N) for j in range(N)]
    left  = [(idx(i,j), idx(i,j-1), -1/h**2) for i in range(N) for j in range(1, N)]
    right = [(idx(i,j), idx(i,j+1), -1/h**2) for i in range(N) for j in range(0, N-1)]
    up    = [(idx(i,j), idx(i-1,j), -1/h**2) for i in range(1, N) for j in range(N)]
    down  = [(idx(i,j), idx(i+1,j), -1/h**2) for i in range(0, N-1) for j in range(N)]
    return diag + left + right + up + down

# ── assemble dense matrix via reduce ─────────────────────────────────────────
def add_entry(A, entry):
    r, c, v = entry
    A[r, c] += v
    return A

A = reduce(add_entry,
           stencil_entries(N, h),
           np.zeros((N * N, N * N)))

# ── right-hand side ────────────────────────────────────────────────────────────
b = np.array([f(xs[i], ys[j]) for i in range(N) for j in range(N)])

# ── solve and reshape ─────────────────────────────────────────────────────────
U = np.linalg.solve(A, b).reshape(N, N)

# ── visualise ─────────────────────────────────────────────────────────────────
X, Y = np.meshgrid(xs, ys)
fig = go.Figure(data=[
    go.Surface(x=X, y=Y, z=U.T, colorscale='Viridis',
               contours=dict(z=dict(show=True, usecolormap=True, highlightcolor="limegreen")))
])
fig.update_layout(
    title='Solution u(x,y) of −Δu = f  [from scratch, N=30]',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='u(x,y)'),
    width=750, height=600,
)
fig.show()
```

```python
import numpy as np
import scipy.sparse as sp
import scipy.sparse.linalg as spla
import plotly.graph_objects as go

# ── problem parameters ────────────────────────────────────────────────────────
N = 80
h = 1 / (N + 1)

# ── source term (same two Gaussian bumps) ─────────────────────────────────────
def f(x, y):
    rho1 = 3.0 * np.exp(-60 * ((x - 0.25)**2 + (y - 0.25)**2))
    rho2 = 1.5 * np.exp(-60 * ((x - 0.70)**2 + (y - 0.65)**2))
    return 2 * np.pi * (rho1 + rho2)

# ── 1-D tridiagonal Laplacian ─────────────────────────────────────────────────
e  = np.ones(N)
T1 = sp.spdiags([-e, 2*e, -e], [-1, 0, 1], N, N, format='csr') / h**2

# ── 2-D Laplacian via Kronecker sum ───────────────────────────────────────────
I  = sp.eye(N, format='csr')
A  = sp.kron(T1, I) + sp.kron(I, T1)          # shape (N², N²), ~5 non-zeros/row

# ── right-hand side ────────────────────────────────────────────────────────────
xs = np.array([(i + 1) * h for i in range(N)])
ys = np.array([(j + 1) * h for j in range(N)])
b  = np.array([f(xs[i], ys[j]) for i in range(N) for j in range(N)])

# ── sparse direct solve ───────────────────────────────────────────────────────
U  = spla.spsolve(A.tocsr(), b).reshape(N, N)

# ── visualise ─────────────────────────────────────────────────────────────────
X, Y = np.meshgrid(xs, ys)
fig = go.Figure(data=[
    go.Surface(x=X, y=Y, z=U.T, colorscale='Viridis',
               contours=dict(z=dict(show=True, usecolormap=True, highlightcolor="limegreen")))
])
fig.update_layout(
    title='Solution u(x,y) of −Δu = f  [scipy.sparse, N=80]',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='u(x,y)'),
    width=750, height=600,
)
fig.show()
```

```python
import numpy as np
import scipy.sparse as sp
import scipy.sparse.linalg as spla
import plotly.graph_objects as go

# ── problem parameters ────────────────────────────────────────────────────────
N = 60
h = 1 / (N + 1)

# ── source term ───────────────────────────────────────────────────────────────
def f(x, y):
    rho1 = 3.0 * np.exp(-60 * ((x - 0.25)**2 + (y - 0.25)**2))
    rho2 = 1.5 * np.exp(-60 * ((x - 0.70)**2 + (y - 0.65)**2))
    return 2 * np.pi * (rho1 + rho2)

xs = np.array([(i + 1) * h for i in range(N)])
ys = np.array([(j + 1) * h for j in range(N)])

# ── solve for u ───────────────────────────────────────────────────────────────
e  = np.ones(N)
T1 = sp.spdiags([-e, 2*e, -e], [-1, 0, 1], N, N, format='csr') / h**2
I  = sp.eye(N, format='csr')
A  = sp.kron(T1, I) + sp.kron(I, T1)

b_flat = np.array([f(xs[i], ys[j]) for i in range(N) for j in range(N)])
U      = spla.spsolve(A.tocsr(), b_flat).reshape(N, N)

# ── recover −Δu numerically (apply the same stiffness matrix A to u) ─────────
neg_lap_u = (A @ U.flatten()).reshape(N, N)

# ── prescribed f on the same grid ─────────────────────────────────────────────
F = np.array([[f(xs[i], ys[j]) for j in range(N)] for i in range(N)])

X, Y = np.meshgrid(xs, ys)

# ── build animation frames (t = 0 → f,  t = 1 → −Δu) ─────────────────────────
n_frames = 30
frames = [
    go.Frame(
        data=[go.Surface(
            x=X, y=Y,
            z=((1 - k/n_frames) * F + (k/n_frames) * neg_lap_u).T,
            colorscale='RdBu', showscale=True,
            cmin=float(F.min()), cmax=float(F.max()),
        )],
        name=str(k),
        layout=go.Layout(
            title_text=(
                f'<b>f(x,y)</b> (source term)' if k == 0
                else f'<b>−Δu(x,y)</b> (recovered from solution)' if k == n_frames
                else f'Morphing f → −Δu   (t = {k/n_frames:.2f})'
            )
        ),
    )
    for k in range(n_frames + 1)
]

# ── slider ────────────────────────────────────────────────────────────────────
slider_steps = [
    dict(
        method='animate',
        args=[[str(k)], dict(mode='immediate',
                             frame=dict(duration=0, redraw=True),
                             transition=dict(duration=0))],
        label=f'{k/n_frames:.2f}',
    )
    for k in range(n_frames + 1)
]

sliders = [dict(
    active=0, steps=slider_steps,
    x=0.05, y=0.02, len=0.90,
    currentvalue=dict(prefix='t = ', visible=True, xanchor='center'),
    transition=dict(duration=0),
)]

# ── figure ────────────────────────────────────────────────────────────────────
fig = go.Figure(
    data=frames[0].data,
    frames=frames,
)

fig.update_layout(
    title='<b>f(x,y)</b> (source term)',
    scene=dict(xaxis_title='x', yaxis_title='y', zaxis_title='',
               camera=dict(eye=dict(x=1.6, y=-1.6, z=1.2))),
    sliders=sliders,
    updatemenus=[dict(
        type='buttons', showactive=False,
        x=0.01, y=0.10, xanchor='left',
        buttons=[
            dict(label='▶ Play',
                 method='animate',
                 args=[None, dict(frame=dict(duration=60, redraw=True),
                                  fromcurrent=True, transition=dict(duration=0))]),
            dict(label='⏸ Pause',
                 method='animate',
                 args=[[None], dict(mode='immediate',
                                    frame=dict(duration=0, redraw=False))]),
        ],
    )],
    width=750, height=680,
)

fig.show()
```
